# NFT Collection Announcement Analyzer: Walkthrough

This notebook runs the [NFT Collection Announcement Analyzer](https://github.com/buycryptopressrelease/nft-collection-announcement-analyzer) on three fictional NFT collections, shows how announcement angles change as the mint date approaches and passes, and prints a full press release outline. Try the browser version at https://buycryptopressrelease.github.io/nft-collection-announcement-analyzer/.

## 1. Load the analyzer and the data

```python
import os, sys, json
import pandas as pd

BASES = ["/kaggle/input/nft-collection-announcement-dataset", "..", "."]
base = next(b for b in BASES if os.path.exists(os.path.join(b, "nft_announcement_analyzer.py")))
data = base if os.path.exists(os.path.join(base, "nft_collections.json")) else os.path.join(base, "data")
sys.path.insert(0, base)
import nft_announcement_analyzer as nft

collections = nft.load(os.path.join(data, "nft_collections.json"))
pd.DataFrame(collections)[["project_name", "blockchain", "collection_type", "total_supply", "minted_supply", "mint_date", "reveal_date"]]
```

## 2. Analyze one collection

```python
report = nft.analyze(collections[0], today="2027-02-01")
print(report["supply"])
print()
for h in report["headlines"]:
    print("-", h)
print()
print("Warnings:", report["warnings"])
```

## 3. How the angles change over time

```python
rows = []
for c in collections:
    for d in ["2026-12-01", "2027-02-01", "2027-03-12", "2027-06-01"]:
        r = nft.analyze(c, d)
        rows.append({"project": c["project_name"], "as_of": d, "supply": r["supply"]["status"],
                     "mint": (r["schedule"]["mint"] or {}).get("status"), "angles": len(r["angles"]),
                     "top headline": r["headlines"][0] if r["headlines"] else ""})
pd.DataFrame(rows)
```

## 4. Full press release outline

```python
print(nft.render(nft.analyze(collections[2], today="2027-02-01"), "md"))
```

## 5. Check a risky claim

```python
test = dict(collections[0], utility=["Guaranteed returns for holders", "Community access"])
print(nft.analyze(test, today="2027-02-01")["warnings"])
```
